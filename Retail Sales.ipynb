```python
import pandas as pd

data = pd.read_csv("sample_superstore.csv", encoding="cp1252")

print("Dataset loaded successfully!")
print("Rows and columns:", data.shape)

data.head()
```

```python
print("Dataset Shape:", data.shape)
```

```python
print("\nColumn Names:")
print(data.columns.tolist())
```

```python
print("\nDataset Information:")
data.info()
```

```python
print("\nMissing Values:")
print(data.isnull().sum())
```

```python
print("\nDuplicate Rows:", data.duplicated().sum())
```

```python
print("\nFirst 5 Records:")
display(data.head())
```

```python
data["Order Date"] = pd.to_datetime(
    data["Order Date"], errors="coerce"
)

data["Ship Date"] = pd.to_datetime(
    data["Ship Date"], errors="coerce"
)
```

```python
print("Date column types:")
print(data[["Order Date", "Ship Date"]].dtypes)

print("\nFirst 5 dates:")
display(data[["Order Date", "Ship Date"]].head())
```

```python
print("Missing Order Dates:", data["Order Date"].isna().sum())
print("Missing Ship Dates:", data["Ship Date"].isna().sum())
```

```python
print("Total rows:", len(data))

print("\nMissing values in important columns:")
print(
    data[["Order Date", "Ship Date", "Sales", "Profit"]]
    .isna()
    .sum()
)

print("\nRows where Order Date is missing:")
display(data[data["Order Date"].isna()].head(10))
```

```python
# Reload the original CSV without converting dates
raw_data = pd.read_csv("sample_superstore.csv", encoding="cp1252")

print("Original Order Date values:")
display(raw_data["Order Date"].head(10))

print("\nOriginal Ship Date values:")
display(raw_data["Ship Date"].head(10))

print("\nOriginal date column data types:")
print(raw_data[["Order Date", "Ship Date"]].dtypes)
```

```python
# Convert dates using the correct DD-MM-YYYY format
data["Order Date"] = pd.to_datetime(
    raw_data["Order Date"],
    format="%d-%m-%Y",
    errors="coerce"
)

data["Ship Date"] = pd.to_datetime(
    raw_data["Ship Date"],
    format="%d-%m-%Y",
    errors="coerce"
)

# Verify the conversion
print("Date column types:")
print(data[["Order Date", "Ship Date"]].dtypes)

print("\nMissing Order Dates:", data["Order Date"].isna().sum())
print("Missing Ship Dates:", data["Ship Date"].isna().sum())

display(data[["Order Date", "Ship Date"]].head(10))
```

```python
numeric_columns = ["Sales", "Profit", "Discount", "Quantity"]

print("Column data types:")
print(data[numeric_columns].dtypes)

print("\nMissing values:")
print(data[numeric_columns].isna().sum())

print("\nSummary statistics:")
display(data[numeric_columns].describe())
```

```python

# 1. Check duplicate records
print("Duplicate rows:", data.duplicated().sum())

# 2. Check discount values
print("\nDiscount range:")
print("Minimum:", data["Discount"].min())
print("Maximum:", data["Discount"].max())

# 3. Check invalid sales and quantities
print("\nSales less than or equal to zero:",
      (data["Sales"] <= 0).sum())

print("Quantity less than or equal to zero:",
      (data["Quantity"] <= 0).sum())

# 4. Count profitable and loss-making records
print("\nProfitable records:",
      (data["Profit"] > 0).sum())

print("Loss-making records:",
      (data["Profit"] < 0).sum())

print("Zero-profit records:",
      (data["Profit"] == 0).sum())
```

```python

# Export the cleaned dataset
data.to_csv("superstore_cleaned.csv", index=False)

print("Cleaned dataset exported successfully!")
print("Total rows:", len(data))
print("Total columns:", len(data.columns))
```
